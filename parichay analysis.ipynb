```python
print("===== Distance Between Consecutive Logins =====")

R = 6371  # Earth radius in KM

lat1 = np.radians(loginDF["PreviousLatitude"])
lon1 = np.radians(loginDF["PreviousLongitude"])

lat2 = np.radians(loginDF["Latitude"])
lon2 = np.radians(loginDF["Longitude"])

dlat = lat2 - lat1
dlon = lon2 - lon1

a = (
    np.sin(dlat / 2) ** 2
    + np.cos(lat1)
    * np.cos(lat2)
    * np.sin(dlon / 2) ** 2
)

c = 2 * np.arctan2(
    np.sqrt(a),
    np.sqrt(1 - a)
)

loginDF["DistanceKM"] = R * c
```

```python
print("===== Feature Engineering =====")

# --------------------------------------------------
# 1. Prepare Login Data
# --------------------------------------------------

loginDF = authLogsDF[
    authLogsDF["EventName"] == "Login"
].copy()

loginDF["LoginTime"] = pd.to_datetime(
    loginDF["TimeStamp"],
    format="%d-%m-%Y %H:%M",
    errors="coerce"
)

loginDF = loginDF.sort_values(
    ["UserName", "LoginTime"]
).reset_index(drop=True)


# --------------------------------------------------
# 2. Previous Login Information
# --------------------------------------------------

loginDF["PreviousLoginTime"] = (
    loginDF.groupby("UserName")["LoginTime"]
    .shift(1)
)

loginDF["PreviousIP"] = (
    loginDF.groupby("UserName")["IPAddress"]
    .shift(1)
)

loginDF["PreviousBrowser"] = (
    loginDF.groupby("UserName")["BrowserID"]
    .shift(1)
)

loginDF["PreviousDevice"] = (
    loginDF.groupby("UserName")["device fingerprinting"]
    .shift(1)
)

loginDF["PreviousState"] = (
    loginDF.groupby("UserName")["State"]
    .shift(1)
)


# --------------------------------------------------
# 3. Time Features
# --------------------------------------------------

loginDF["TimeDifferenceMinutes"] = (
    loginDF["LoginTime"] -
    loginDF["PreviousLoginTime"]
).dt.total_seconds() / 60

loginDF["LoginHour"] = loginDF["LoginTime"].dt.hour

loginDF["IsNightLogin"] = (
    (loginDF["LoginHour"] < 6) |
    (loginDF["LoginHour"] >= 22)
).astype(int)


# --------------------------------------------------
# 4. Behavior Change Features
# --------------------------------------------------

loginDF["IPChanged"] = (
    loginDF["IPAddress"] != loginDF["PreviousIP"]
).astype(int)

loginDF["BrowserChanged"] = (
    loginDF["BrowserID"] != loginDF["PreviousBrowser"]
).astype(int)

loginDF["DeviceChanged"] = (
    loginDF["device fingerprinting"] !=
    loginDF["PreviousDevice"]
).astype(int)

loginDF["StateChanged"] = (
    loginDF["State"] != loginDF["PreviousState"]
).astype(int)

# First login has no previous login
firstLogin = loginDF["PreviousLoginTime"].isna()

changeColumns = [
    "IPChanged",
    "BrowserChanged",
    "DeviceChanged",
    "StateChanged"
]

loginDF.loc[firstLogin, changeColumns] = 0


# --------------------------------------------------
# 5. Distance Between Previous and Current Login
# --------------------------------------------------

R = 6371

lat1 = np.radians(
    loginDF.groupby("UserName")["Latitude"].shift(1)
)

lon1 = np.radians(
    loginDF.groupby("UserName")["Longitude"].shift(1)
)

lat2 = np.radians(loginDF["Latitude"])
lon2 = np.radians(loginDF["Longitude"])

dlat = lat2 - lat1
dlon = lon2 - lon1

a = (
    np.sin(dlat / 2) ** 2
    +
    np.cos(lat1)
    * np.cos(lat2)
    * np.sin(dlon / 2) ** 2
)

c = 2 * np.arctan2(
    np.sqrt(a),
    np.sqrt(1 - a)
)

loginDF["DistanceKM"] = R * c

loginDF.loc[firstLogin, "DistanceKM"] = 0


# --------------------------------------------------
# 6. Travel Speed
# --------------------------------------------------

loginDF["TimeDifferenceHours"] = (
    loginDF["TimeDifferenceMinutes"] / 60
)

loginDF["TravelSpeedKMH"] = np.where(
    loginDF["TimeDifferenceHours"] > 0,
    loginDF["DistanceKM"] /
    loginDF["TimeDifferenceHours"],
    0
)

loginDF.loc[firstLogin, "TravelSpeedKMH"] = 0


# --------------------------------------------------
# 7. Current Login Security Features
# --------------------------------------------------

loginDF["LoginFailure"] = (
    loginDF["EventType"] == "Failure"
).astype(int)

loginDF["LoginWarning"] = (
    loginDF["EventType"] == "Warning"
).astype(int)

loginDF["LoginAlertFlag"] = (
    loginDF["LoginAlert"] == "Yes"
).astype(int)

loginDF["GeoFencingDisabled"] = (
    loginDF["WithGeoFencing"] == "No"
).astype(int)

loginDF["UnknownDevice"] = (
    loginDF["RememberDevice"] == "No"
).astype(int)

loginDF["VPNFlag"] = (
    loginDF["IsVPN"] == "Yes"
).astype(int)

loginDF["ProxyFlag"] = (
    loginDF["IsProxy"] == "Yes"
).astype(int)


# --------------------------------------------------
# 8. User History
# --------------------------------------------------

loginDF["PreviousLoginCount"] = (
    loginDF.groupby("UserName")
    .cumcount()
)

loginDF["IsNewIP"] = (
    ~loginDF.duplicated(
        ["UserName", "IPAddress"],
        keep="first"
    )
).astype(int)

loginDF["IsNewBrowser"] = (
    ~loginDF.duplicated(
        ["UserName", "BrowserID"],
        keep="first"
    )
).astype(int)

loginDF["IsNewDevice"] = (
    ~loginDF.duplicated(
        ["UserName", "device fingerprinting"],
        keep="first"
    )
).astype(int)


# --------------------------------------------------
# 9. Select Final ML Features
# --------------------------------------------------

featureColumns = [
    "TimeDifferenceMinutes",
    "LoginHour",
    "IsNightLogin",

    "IPChanged",
    "BrowserChanged",
    "DeviceChanged",
    "StateChanged",

    "DistanceKM",
    "TravelSpeedKMH",

    "LoginFailure",
    "LoginWarning",
    "LoginAlertFlag",
    "GeoFencingDisabled",
    "UnknownDevice",
    "VPNFlag",
    "ProxyFlag",
    "IPReputationScore",

    "PreviousLoginCount",
    "IsNewIP",
    "IsNewBrowser",
    "IsNewDevice"
]


# --------------------------------------------------
# 10. Create ML Dataset
# --------------------------------------------------

mlFeaturesDF = loginDF[
    ["UserName", "LoginTime"] + featureColumns
].copy()


# --------------------------------------------------
# 11. Handle NaN / Infinite Values
# --------------------------------------------------

mlFeaturesDF[featureColumns] = (
    mlFeaturesDF[featureColumns]
    .replace([np.inf, -np.inf], np.nan)
    .fillna(0)
)


print("\n===== Feature Engineering Completed =====")

print("Original Logs:", len(authLogsDF))
print("Login Observations:", len(mlFeaturesDF))
print("Number of ML Features:", len(featureColumns))

print("\nML Dataset Shape:")
print(mlFeaturesDF.shape)

print("\nFeature Columns:")
print(featureColumns)

print("\nFirst 5 Rows:")
display(mlFeaturesDF.head())
```

```python
print("===== Feature Quality Check =====")

X = mlFeaturesDF[featureColumns].copy()

print("Shape:", X.shape)

print("\nMissing Values:")
print(X.isnull().sum().sum())

print("\nInfinite Values:")
print(np.isinf(X).sum().sum())

print("\nData Types:")
print(X.dtypes.value_counts())

print("\nConstant Features:")
constantFeatures = [
    col for col in X.columns
    if X[col].nunique() <= 1
]

print(constantFeatures)

print("\nFeature Statistics:")
display(X.describe().T)
```

```python
print("===== Feature Correlation Analysis =====")

correlationDF = X.corr()

print(
    correlationDF.round(2)
)
```

```python
highCorrelation = (
    correlationDF
    .where(
        np.triu(
            np.ones(correlationDF.shape),
            k=1
        ).astype(bool)
    )
    .stack()
    .reset_index()
)

highCorrelation.columns = [
    "Feature1",
    "Feature2",
    "Correlation"
]

highCorrelation = highCorrelation[
    highCorrelation["Correlation"].abs() >= 0.80
].sort_values(
    "Correlation",
    key=lambda x: x.abs(),
    ascending=False
)

print("\n===== Highly Correlated Features =====")
print(highCorrelation)
```

```python
print("===== Model Preparation =====")

# Only ML features
X = mlFeaturesDF[featureColumns].copy()

print("X Shape:", X.shape)

print("\nData Types:")
print(X.dtypes)

print("\nMissing Values:", X.isnull().sum().sum())

print("\nInfinite Values:", np.isinf(X).sum().sum())

print("\nFirst 5 Rows:")
display(X.head())
```

```python
print("===== Isolation Forest Training =====")

model = IsolationForest(
    n_estimators=200,
    contamination=0.05,
    random_state=42
)

model.fit(X)

print("Model training completed.")
```

```python
print("===== Model Prediction =====")

mlFeaturesDF["Prediction"] = model.predict(X)

print(
    mlFeaturesDF[
        ["UserName", "LoginTime", "Prediction"]
    ].head(20)
)
```

```python
print("===== Anomaly Score =====")

mlFeaturesDF["AnomalyScore"] = (
    model.decision_function(X)
)

print(
    mlFeaturesDF[
        [
            "UserName",
            "LoginTime",
            "Prediction",
            "AnomalyScore"
        ]
    ]
    .head(20)
)
```

```python
print("===== Anomaly Validation =====")

validationDF = mlFeaturesDF[
    mlFeaturesDF["Prediction"] == -1
].copy()

# Independent suspicious signals

validationDF["ImpossibleTravel"] = (
    (validationDF["DistanceKM"] >= 500) &
    (validationDF["TimeDifferenceMinutes"] <= 60)
).astype(int)

validationDF["MultipleChanges"] = (
    validationDF[
        [
            "IPChanged",
            "BrowserChanged",
            "DeviceChanged",
            "StateChanged"
        ]
    ].sum(axis=1) >= 2
).astype(int)

validationDF["FailedLoginUnknownDevice"] = (
    (validationDF["LoginFailure"] == 1) &
    (validationDF["UnknownDevice"] == 1)
).astype(int)

validationDF["NetworkRisk"] = (
    (
        (validationDF["VPNFlag"] == 1) |
        (validationDF["ProxyFlag"] == 1)
    ) &
    (validationDF["IPReputationScore"] < 50)
).astype(int)

validationDF["AlertSignal"] = (
    validationDF["LoginAlertFlag"] == 1
).astype(int)


# Count suspicious signals
validationDF["SuspiciousSignalCount"] = (
    validationDF[
        [
            "ImpossibleTravel",
            "MultipleChanges",
            "FailedLoginUnknownDevice",
            "NetworkRisk",
            "AlertSignal"
        ]
    ].sum(axis=1)
)


print("Total Model Anomalies:", len(validationDF))

print("\n===== Validation Summary =====")

print(
    validationDF["SuspiciousSignalCount"]
    .value_counts()
    .sort_index()
)

print("\n===== Signal Coverage =====")

print(
    validationDF[
        [
            "ImpossibleTravel",
            "MultipleChanges",
            "FailedLoginUnknownDevice",
            "NetworkRisk",
            "AlertSignal"
        ]
    ].sum()
)
```

```python
print("===== Risk Categorization =====")

riskDF = mlFeaturesDF.copy()

# --------------------------------------------------
# 1. Calculate Suspicious Signals
# --------------------------------------------------

riskDF["ImpossibleTravel"] = (
    (riskDF["DistanceKM"] >= 500) &
    (riskDF["TimeDifferenceMinutes"] <= 60)
).astype(int)

riskDF["MultipleChanges"] = (
    riskDF[
        [
            "IPChanged",
            "BrowserChanged",
            "DeviceChanged",
            "StateChanged"
        ]
    ].sum(axis=1) >= 2
).astype(int)

riskDF["FailedLoginUnknownDevice"] = (
    (riskDF["LoginFailure"] == 1) &
    (riskDF["UnknownDevice"] == 1)
).astype(int)

riskDF["NetworkRisk"] = (
    (
        (riskDF["VPNFlag"] == 1) |
        (riskDF["ProxyFlag"] == 1)
    ) &
    (riskDF["IPReputationScore"] < 50)
).astype(int)

riskDF["AlertSignal"] = (
    riskDF["LoginAlertFlag"] == 1
).astype(int)


# --------------------------------------------------
# 2. Total Suspicious Signals
# --------------------------------------------------

riskDF["SuspiciousSignalCount"] = (
    riskDF[
        [
            "ImpossibleTravel",
            "MultipleChanges",
            "FailedLoginUnknownDevice",
            "NetworkRisk",
            "AlertSignal"
        ]
    ].sum(axis=1)
)


# --------------------------------------------------
# 3. Risk Level
# --------------------------------------------------

riskDF["RiskLevel"] = np.select(
    [
        (riskDF["Prediction"] == -1) &
        (riskDF["SuspiciousSignalCount"] >= 3),

        (riskDF["Prediction"] == -1) &
        (riskDF["SuspiciousSignalCount"] >= 2),

        (riskDF["Prediction"] == -1) &
        (riskDF["SuspiciousSignalCount"] >= 1)
    ],
    [
        "High",
        "Medium",
        "Low"
    ],
    default="Normal"
)


# --------------------------------------------------
# 4. Summary
# --------------------------------------------------

print("\n===== Risk Level Distribution =====")

print(
    riskDF["RiskLevel"]
    .value_counts()
)


# --------------------------------------------------
# 5. Suspicious Login Report
# --------------------------------------------------

suspiciousReportDF = (
    riskDF[
        riskDF["RiskLevel"] != "Normal"
    ]
    .sort_values(
        ["RiskLevel", "AnomalyScore"],
        ascending=[True, True]
    )
)

print("\n===== Suspicious Login Report =====")

print(
    suspiciousReportDF[
        [
            "UserName",
            "LoginTime",
            "Prediction",
            "AnomalyScore",
            "SuspiciousSignalCount",
            "RiskLevel",
            "DistanceKM",
            "TravelSpeedKMH",
            "IPChanged",
            "BrowserChanged",
            "DeviceChanged",
            "StateChanged",
            "LoginFailure",
            "LoginAlertFlag",
           
        ]
    ]
    .head(20)
)
```

```python
print("===== Feature Engineering on Unseen Data =====")

# --------------------------------------------------
# 1. Prepare Login Data
# --------------------------------------------------

testLoginDF = testDF[
    testDF["EventName"] == "Login"
].copy()

testLoginDF["LoginTime"] = pd.to_datetime(
    testLoginDF["TimeStamp"],
    format="%d-%m-%Y %H:%M",
    errors="coerce"
)

testLoginDF = testLoginDF.sort_values(
    ["UserName", "LoginTime"]
).reset_index(drop=True)


# --------------------------------------------------
# 2. Previous Login Information
# --------------------------------------------------

testLoginDF["PreviousLoginTime"] = (
    testLoginDF.groupby("UserName")["LoginTime"]
    .shift(1)
)

testLoginDF["PreviousIP"] = (
    testLoginDF.groupby("UserName")["IPAddress"]
    .shift(1)
)

testLoginDF["PreviousBrowser"] = (
    testLoginDF.groupby("UserName")["BrowserID"]
    .shift(1)
)

testLoginDF["PreviousDevice"] = (
    testLoginDF.groupby("UserName")["device fingerprinting"]
    .shift(1)
)

testLoginDF["PreviousState"] = (
    testLoginDF.groupby("UserName")["State"]
    .shift(1)
)


# --------------------------------------------------
# 3. Time Features
# --------------------------------------------------

testLoginDF["TimeDifferenceMinutes"] = (
    testLoginDF["LoginTime"]
    - testLoginDF["PreviousLoginTime"]
).dt.total_seconds() / 60

testLoginDF["LoginHour"] = (
    testLoginDF["LoginTime"].dt.hour
)

testLoginDF["IsNightLogin"] = (
    (testLoginDF["LoginHour"] < 6) |
    (testLoginDF["LoginHour"] >= 22)
).astype(int)


# --------------------------------------------------
# 4. Behavior Change Features
# --------------------------------------------------

testLoginDF["IPChanged"] = (
    testLoginDF["IPAddress"] !=
    testLoginDF["PreviousIP"]
).astype(int)

testLoginDF["BrowserChanged"] = (
    testLoginDF["BrowserID"] !=
    testLoginDF["PreviousBrowser"]
).astype(int)

testLoginDF["DeviceChanged"] = (
    testLoginDF["device fingerprinting"] !=
    testLoginDF["PreviousDevice"]
).astype(int)

testLoginDF["StateChanged"] = (
    testLoginDF["State"] !=
    testLoginDF["PreviousState"]
).astype(int)


# First login of each user has no previous login
firstLogin = (
    testLoginDF["PreviousLoginTime"].isna()
)

changeColumns = [
    "IPChanged",
    "BrowserChanged",
    "DeviceChanged",
    "StateChanged"
]

testLoginDF.loc[
    firstLogin,
    changeColumns
] = 0


# --------------------------------------------------
# 5. Distance
# --------------------------------------------------

R = 6371

lat1 = np.radians(
    testLoginDF.groupby("UserName")["Latitude"].shift(1)
)

lon1 = np.radians(
    testLoginDF.groupby("UserName")["Longitude"].shift(1)
)

lat2 = np.radians(
    testLoginDF["Latitude"]
)

lon2 = np.radians(
    testLoginDF["Longitude"]
)

dlat = lat2 - lat1
dlon = lon2 - lon1

a = (
    np.sin(dlat / 2) ** 2
    +
    np.cos(lat1)
    * np.cos(lat2)
    * np.sin(dlon / 2) ** 2
)

c = 2 * np.arctan2(
    np.sqrt(a),
    np.sqrt(1 - a)
)

testLoginDF["DistanceKM"] = R * c

testLoginDF.loc[
    firstLogin,
    "DistanceKM"
] = 0


# --------------------------------------------------
# 6. Travel Speed
# --------------------------------------------------

testLoginDF["TimeDifferenceHours"] = (
    testLoginDF["TimeDifferenceMinutes"] / 60
)

testLoginDF["TravelSpeedKMH"] = np.where(
    testLoginDF["TimeDifferenceHours"] > 0,
    testLoginDF["DistanceKM"] /
    testLoginDF["TimeDifferenceHours"],
    0
)

testLoginDF.loc[
    firstLogin,
    "TravelSpeedKMH"
] = 0


# --------------------------------------------------
# 7. Security Features
# --------------------------------------------------

testLoginDF["LoginFailure"] = (
    testLoginDF["EventType"] == "Failure"
).astype(int)

testLoginDF["LoginWarning"] = (
    testLoginDF["EventType"] == "Warning"
).astype(int)

testLoginDF["LoginAlertFlag"] = (
    testLoginDF["LoginAlert"] == "Yes"
).astype(int)

testLoginDF["GeoFencingDisabled"] = (
    testLoginDF["WithGeoFencing"] == "No"
).astype(int)

testLoginDF["UnknownDevice"] = (
    testLoginDF["RememberDevice"] == "No"
).astype(int)

testLoginDF["VPNFlag"] = (
    testLoginDF["IsVPN"] == "Yes"
).astype(int)

testLoginDF["ProxyFlag"] = (
    testLoginDF["IsProxy"] == "Yes"
).astype(int)


# --------------------------------------------------
# 8. User History
# --------------------------------------------------

testLoginDF["PreviousLoginCount"] = (
    testLoginDF.groupby("UserName")
    .cumcount()
)

testLoginDF["IsNewIP"] = (
    ~testLoginDF.duplicated(
        ["UserName", "IPAddress"],
        keep="first"
    )
).astype(int)

testLoginDF["IsNewBrowser"] = (
    ~testLoginDF.duplicated(
        ["UserName", "BrowserID"],
        keep="first"
    )
).astype(int)

testLoginDF["IsNewDevice"] = (
    ~testLoginDF.duplicated(
        ["UserName", "device fingerprinting"],
        keep="first"
    )
).astype(int)


# --------------------------------------------------
# 9. Create Test ML Dataset
# --------------------------------------------------

testMLDF = testLoginDF[
    ["UserName", "LoginTime", "ActualLabel"]
    + featureColumns
].copy()


# --------------------------------------------------
# 10. Clean Features
# --------------------------------------------------

testMLDF[featureColumns] = (
    testMLDF[featureColumns]
    .replace([np.inf, -np.inf], np.nan)
    .fillna(0)
)


print("\n===== Unseen Feature Engineering Completed =====")

print("Test Login Observations:", len(testMLDF))

print(
    "Number of ML Features:",
    len(featureColumns)
)

print(
    "\nTest ML Dataset Shape:",
    testMLDF.shape
)

print("\nFeature Columns:")
print(featureColumns)

print("\nFirst 5 Rows:")
display(testMLDF.head())
```

```python
print("===== User Brute Force Detection =====")

# Filter failed login events
failedLoginDF = authLogsDF[
    (authLogsDF["EventName"] == "Login") &
    (authLogsDF["EventType"] == "Failure")
].copy()

# Convert timestamp
failedLoginDF["LoginTime"] = pd.to_datetime(
    failedLoginDF["TimeStamp"],
    format="%d-%m-%Y %H:%M",
    errors="coerce"
)

# Sort by User and LoginTime
failedLoginDF = failedLoginDF.sort_values(
    ["UserName", "LoginTime"]
)

# Count failed attempts within the previous 10 minutes
def failed_attempts(group):
    group = group.copy()

    group["FailedAttempts"] = (
        group.set_index("LoginTime")
             .rolling("10min")
             .count()["UserName"]
             .values
    )

    return group

userBruteForceDF = (
    failedLoginDF
    .groupby("UserName", group_keys=False)
    .apply(failed_attempts)
)

# Assign Risk Level
conditions = [
    userBruteForceDF["FailedAttempts"] >= 10,
    userBruteForceDF["FailedAttempts"] >= 5,
    userBruteForceDF["FailedAttempts"] >= 3
]

choices = [
    "Critical",
    "High",
    "Medium"
]

userBruteForceDF["RiskLevel"] = np.select(
    conditions,
    choices,
    default="Low"
)

# Keep High and Critical only (>=5 attempts)
userBruteForceDF = (
    userBruteForceDF[
        userBruteForceDF["FailedAttempts"] >= 5
    ][
        [
            "UserName",
            "IPAddress",
            "LoginTime",
            "FailedAttempts",
            "RiskLevel"
        ]
    ]
    .drop_duplicates()
)

print(userBruteForceDF)
```

```python
print("===== Core Feature Engineering =====")

# --------------------------------------------------
# 1. Prepare Login Data
# --------------------------------------------------

loginDF = authLogsDF[
    authLogsDF["EventName"] == "Login"
].copy()

loginDF["LoginTime"] = pd.to_datetime(
    loginDF["TimeStamp"],
    format="%d-%m-%Y %H:%M",
    errors="coerce"
)

loginDF = (
    loginDF
    .sort_values(["UserName", "LoginTime"])
    .reset_index(drop=True)
)


# --------------------------------------------------
# 2. Previous Login Information
# --------------------------------------------------

loginDF["PreviousLoginTime"] = (
    loginDF
    .groupby("UserName")["LoginTime"]
    .shift(1)
)

loginDF["PreviousIP"] = (
    loginDF
    .groupby("UserName")["IPAddress"]
    .shift(1)
)

loginDF["PreviousBrowser"] = (
    loginDF
    .groupby("UserName")["BrowserID"]
    .shift(1)
)

loginDF["PreviousState"] = (
    loginDF
    .groupby("UserName")["State"]
    .shift(1)
)

loginDF["PreviousLatitude"] = (
    loginDF
    .groupby("UserName")["Latitude"]
    .shift(1)
)

loginDF["PreviousLongitude"] = (
    loginDF
    .groupby("UserName")["Longitude"]
    .shift(1)
)


# --------------------------------------------------
# 3. Time Difference
# --------------------------------------------------

loginDF["TimeDifferenceMinutes"] = (
    loginDF["LoginTime"]
    -
    loginDF["PreviousLoginTime"]
).dt.total_seconds() / 60


# --------------------------------------------------
# 4. IP Change
# --------------------------------------------------

loginDF["IPChanged"] = (
    loginDF["IPAddress"]
    != loginDF["PreviousIP"]
).astype(int)


# --------------------------------------------------
# 5. Browser Change
# --------------------------------------------------

loginDF["BrowserChanged"] = (
    loginDF["BrowserID"]
    != loginDF["PreviousBrowser"]
).astype(int)


# --------------------------------------------------
# 6. State Change
# --------------------------------------------------

loginDF["StateChanged"] = (
    loginDF["State"]
    != loginDF["PreviousState"]
).astype(int)


# --------------------------------------------------
# 7. Login Failure
# --------------------------------------------------

loginDF["LoginFailure"] = (
    loginDF["EventType"] == "Failure"
).astype(int)


# First login has no previous login
firstLogin = (
    loginDF["PreviousLoginTime"].isna()
)

loginDF.loc[
    firstLogin,
    [
        "IPChanged",
        "BrowserChanged",
        "StateChanged"
    ]
] = 0


# --------------------------------------------------
# 8. Distance Between Logins
# --------------------------------------------------

R = 6371

lat1 = np.radians(
    loginDF["PreviousLatitude"]
)

lon1 = np.radians(
    loginDF["PreviousLongitude"]
)

lat2 = np.radians(
    loginDF["Latitude"]
)

lon2 = np.radians(
    loginDF["Longitude"]
)

dlat = lat2 - lat1
dlon = lon2 - lon1

a = (
    np.sin(dlat / 2) ** 2
    +
    np.cos(lat1)
    * np.cos(lat2)
    * np.sin(dlon / 2) ** 2
)

c = 2 * np.arctan2(
    np.sqrt(a),
    np.sqrt(1 - a)
)

loginDF["DistanceKM"] = R * c

loginDF.loc[
    firstLogin,
    "DistanceKM"
] = 0


# --------------------------------------------------
# 9. Travel Speed
# --------------------------------------------------

loginDF["TimeDifferenceHours"] = (
    loginDF["TimeDifferenceMinutes"] / 60
)

loginDF["TravelSpeedKMH"] = np.where(
    loginDF["TimeDifferenceHours"] > 0,
    loginDF["DistanceKM"]
    /
    loginDF["TimeDifferenceHours"],
    0
)

loginDF.loc[
    firstLogin,
    "TravelSpeedKMH"
] = 0


# --------------------------------------------------
# 10. Final ML Features
# --------------------------------------------------

coreFeatureColumns = [
    "IPChanged",
    "BrowserChanged",
    "StateChanged",
    "LoginFailure",
    "TimeDifferenceMinutes",
    "DistanceKM",
    "TravelSpeedKMH"
]


# --------------------------------------------------
# 11. Create ML Dataset
# --------------------------------------------------

coreMLDF = loginDF[
    ["UserName", "LoginTime"] +
    coreFeatureColumns
].copy()


# --------------------------------------------------
# 12. Handle Invalid Values
# --------------------------------------------------

coreMLDF[coreFeatureColumns] = (
    coreMLDF[coreFeatureColumns]
    .replace(
        [np.inf, -np.inf],
        np.nan
    )
    .fillna(0)
)


print("\n===== Core Feature Engineering Completed =====")

print(
    "Login Observations:",
    len(coreMLDF)
)

print(
    "Number of ML Features:",
    len(coreFeatureColumns)
)

print("\nFeature Columns:")
print(coreFeatureColumns)

print("\nFirst 5 Rows:")
display(coreMLDF.head())
```
